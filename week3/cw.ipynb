```python
import numpy as np
```

```python
y = [0.0, 0.0, 0.0, 1.0]
y_pred = [0.1, 0.2, 0.6, 0.1]
```

```python
def cross_entropy(y, y_pred, eps=1e-12):
    y = np.asarray(y, dtype=float)
    p = np.asarray(y_pred, dtype=float)
    return -(y * np.log(p) + (1.0 - y) * np.log(1.0 - p))
```

```python
print(cross_entropy(y, y_pred))
```
